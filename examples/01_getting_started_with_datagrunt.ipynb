## What is Datagrunt?

Datagrunt is a Python library designed to simplify the way you work with CSV files. It provides:

- ⚡ **Seamless Data Processing** - Leverages DuckDB and Polars for high performance
- 🔄 **Flexible Transformation** - Easy conversion between formats
- 🐍 **Pythonic API** - Clean and intuitive interface

## 1. Basic Usage - Reading CSV Files

Let's start with the basics. Datagrunt makes it incredibly easy to read CSV files!

```python
# standard library imports
import os
import random

# third-party imports
from datagrunt import CSVReader
import polars as pl

# First, let's create a sample CSV file to work with
sample_data = pl.DataFrame({
    'product_id': ['P001', 'P002', 'P003', 'P004', 'P005'],
    'product_name': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Headphones'],
    'category': ['Electronics', 'Electronics', 'Electronics', 'Electronics', 'Electronics'],
    'price': [899.99, 29.99, 79.99, 299.99, 149.99],
    'quantity': [50, 200, 150, 75, 100],
    'rating': [4.5, 4.2, 4.8, 4.6, 4.3]
})

# Save to CSV

# Method 3: Check first, then create
if not os.path.exists('data'):
    os.makedirs('data')
    print("Directory 'data' created successfully")
else:
    print("Directory 'data' already exists")

csv_comma_file = 'data/products.csv'
sample_data.write_csv(csv_comma_file)
print("✅ Sample CSV file created!")
```

### Loading Data with Datagrunt

Now let's load our CSV file using Datagrunt. You can choose between two powerful engines:
- **DuckDB**: Great for SQL queries and analytical workloads
- **Polars**: Lightning-fast DataFrame operations (default)

```python
# Load CSV using DuckDB engine
dg_duckdb = CSVReader(csv_comma_file, engine='duckdb')

# Load CSV using Polars engine (default)
dg_polars = CSVReader(csv_comma_file, engine='polars')

print("✅ CSV loaded with both engines!")
```

## 2. Exploring Your Data

### Get a Sample of Your Data

The `get_sample()` method gives you a quick peek at your data structure:

```python
# Get a sample of the data
print("📊 Data Sample (DuckDB Engine):")
print("=" * 50)
dg_duckdb.get_sample()
```

## 3. Working with Different Delimiters

One of Datagrunt's superpowers is **automatic delimiter detection**! Let's test this with different file formats:

```python
# Create files with different delimiters
# Semicolon-separated
csv_semicolon_file = 'data/products_semicolon.csv'
sample_data.write_csv(csv_semicolon_file, separator=';')

# Tab-separated
csv_tab_file = 'data/products_tab.tsv'
sample_data.write_csv(csv_tab_file, separator='\t')

# Pipe-separated
csv_pipe_file = 'data/products_pipe.csv'
sample_data.write_csv(csv_pipe_file, separator='|')

print("✅ Files with different delimiters created!")
```

```python
# Datagrunt automatically detects the delimiter!
delimiters = {
    'Semicolon': csv_semicolon_file,
    'Tab': csv_tab_file,
    'Pipe': csv_pipe_file
}

for delimiter_type, filename in delimiters.items():
    dg = CSVReader(filename, engine='duckdb')
    print(f"\n📁 {delimiter_type}-separated file:")
    print(f"Datagrunt successfully loaded {filename}!")
    # Show first few rows to confirm
    sample = dg.get_sample()
    print("✅ Delimiter automatically detected and data loaded correctly!")
```

## 4. Converting Between Formats

Datagrunt makes it easy to convert your CSV data into different formats. Let's explore the conversion capabilities:

```python
# Load our products CSV
dg = CSVReader(csv_comma_file, engine='duckdb')

# Simple query to get all data
query = f"SELECT * FROM {dg.db_table}"
result = dg.query_data(query)

# Convert to Polars DataFrame
polars_df = result.pl()
print("🐻‍❄️ Polars DataFrame:")
print(polars_df)
print(f"\nType: {type(polars_df)}")
```

```python
# You can also convert query results to other formats
# Let's run a simple aggregation query
agg_query = f"""
SELECT
    category,
    COUNT(*) as product_count,
    AVG(CAST(price AS float)) as avg_price,
    SUM(CAST(quantity AS integer)) as total_quantity
FROM {dg.db_table}
GROUP BY category
"""

result = dg.query_data(agg_query).pl()
# aggregated_df = result.pl()

print("📊 Aggregated Results:")
print(result)
```

## 5. Handling Large Files

Let's create a larger dataset to demonstrate Datagrunt's efficiency with bigger files:

```python
# Create a larger dataset
random.seed(42)
n_rows = 100000
categories = ['Electronics', 'Clothing', 'Books', 'Home & Garden', 'Sports']
brands = ['BrandA', 'BrandB', 'BrandC', 'BrandD', 'BrandE']

# Generate data using Python's random module
large_data = pl.DataFrame({
    'order_id': [f'ORD{i:06d}' for i in range(n_rows)],
    'product_name': [f'Product_{i % 1000}' for i in range(n_rows)],
    'category': [random.choice(categories) for _ in range(n_rows)],
    'brand': [random.choice(brands) for _ in range(n_rows)],
    'price': [round(random.uniform(10, 1000), 2) for _ in range(n_rows)],
    'quantity': [random.randint(1, 99) for _ in range(n_rows)],
    'discount': [round(random.uniform(0, 0.5), 2) for _ in range(n_rows)],
    'rating': [round(random.uniform(1, 5), 1) for _ in range(n_rows)]
})

# Save to CSV
large_csv_file = 'data/large_orders.csv'
large_data.write_csv(large_csv_file)
print(f"✅ Large dataset created with {n_rows:,} rows!")
print(f"📁 File size: {large_data.estimated_size() / 1024**2:.2f} MB (in memory)")
```

```python
# Load the large dataset with Datagrunt
import time

# Time the loading process
start_time = time.time()
dg_large = CSVReader(large_csv_file, engine='duckdb')
load_time = time.time() - start_time

print(f"⚡ Loaded {n_rows:,} rows in {load_time:.2f} seconds!")
print("\n📊 Sample of large dataset:")
dg_large.get_sample()
```

## 6. Performance Comparison: DuckDB vs Polars Engines

Let's compare the performance of both engines on our large dataset:

```python
# Complex analytical query
complex_query = f"""
WITH
casttypes AS (
    SELECT
        category,
        brand,
        CAST(price AS float) as price,
        CAST(quantity AS integer) as quantity,
        CAST(discount AS float) as discount,
        CAST(rating AS float) as rating
    FROM {dg_large.db_table}
),
sales_summary AS (
    SELECT
        category,
        brand,
        COUNT(*) as order_count,
        SUM(price * quantity * (1 - discount)) as total_revenue,
        AVG(rating) as avg_rating,
        MAX(price) as max_price,
        MIN(price) as min_price
    FROM casttypes
    WHERE price > 0 AND quantity > 0
    GROUP BY category, brand
)
SELECT
    category,
    brand,
    order_count,
    ROUND(total_revenue, 2) as total_revenue,
    ROUND(avg_rating, 2) as avg_rating,
    ROUND(max_price - min_price, 2) as price_range
FROM sales_summary
WHERE order_count > 100
ORDER BY total_revenue DESC
LIMIT 10
"""

# Test with DuckDB engine
start_time = time.time()
result_duckdb = dg_large.query_data(complex_query)
df_duckdb = result_duckdb.pl()
duckdb_time = time.time() - start_time

print("🦆 DuckDB Engine Results:")
print(df_duckdb)
print(f"\n⏱️ Execution time: {duckdb_time:.3f} seconds")
```

## 7. Tips and Best Practices 💡

### When to use DuckDB engine:
- Complex SQL queries with multiple joins
- Analytical workloads with aggregations
- When you're comfortable with SQL

### When to use Polars engine:
- Simple data transformations
- When you need the fastest possible performance
- When working with the Polars ecosystem

### General tips:
1. **Let Datagrunt handle delimiters** - Don't worry about specifying them!
2. **Use `get_sample()`** - Always preview your data first
3. **Leverage SQL** - DuckDB's SQL capabilities are powerful
4. **Convert wisely** - Use `.pl()` for Polars DataFrames when needed
